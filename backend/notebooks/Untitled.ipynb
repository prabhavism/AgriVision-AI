```python
def flag_outliers_iqr(df, group_col, value_col, factor=3):
    def bounds(x):
        q1, q3 = x.quantile(0.25), x.quantile(0.75)
        iqr = q3 - q1
        return q1 - factor*iqr, q3 + factor*iqr
    
    outlier_mask = pd.Series(False, index=df.index)
    for crop, group in df.groupby(group_col):
        low, high = bounds(group[value_col])
        mask = (df[group_col] == crop) & ((df[value_col] < low) | (df[value_col] > high))
        outlier_mask |= mask
    return outlier_mask

outliers = flag_outliers_iqr(df1, 'Crop', 'Yield_kg_per_ha')
print("Number of outlier rows:", outliers.sum())
print(df1[outliers][['Crop','State Name','Year','Yield_kg_per_ha']])
```

```python
realistic_max = {
    'rice': 10000,
    'maize': 15000,
    'chickpea': 3000,
    'cotton': 3000
}

def is_impossible(row):
    return row['Yield_kg_per_ha'] > realistic_max[row['Crop']]

impossible_mask = df1.apply(is_impossible, axis=1)
print("Impossible rows:", impossible_mask.sum())
print(df1[impossible_mask][['Crop','State Name','Year','Yield_kg_per_ha']])
```

```python
drop_indices = [48486, 48786, 31641]
df1_clean = df1.drop(index=drop_indices).reset_index(drop=True)
print("Original rows:", len(df1))
print("Cleaned rows:", len(df1_clean))
```

```python
df1_clean.columns = (
    df1_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('%', 'pct')
)
print(df1_clean.columns.tolist())
```

```python
df1_clean.to_csv("../data/processed/dataset1_clean.csv", index=False)
print("Saved dataset1_clean.csv")
```
