```python
def flag_outliers_iqr(df, group_col, value_col, factor=3):
    def bounds(x):
        q1, q3 = x.quantile(0.25), x.quantile(0.75)
        iqr = q3 - q1
        return q1 - factor*iqr, q3 + factor*iqr
    
    outlier_mask = pd.Series(False, index=df.index)
    for crop, group in df.groupby(group_col):
        low, high = bounds(group[value_col])
        mask = (df[group_col] == crop) & ((df[value_col] < low) | (df[value_col] > high))
        outlier_mask |= mask
    return outlier_mask

outliers = flag_outliers_iqr(df1, 'Crop', 'Yield_kg_per_ha')
print("Number of outlier rows:", outliers.sum())
print(df1[outliers][['Crop','State Name','Year','Yield_kg_per_ha']])
```

```python
import requests
import time

def geocode_district(district, state, country="India"):
    url = "https://nominatim.openstreetmap.org/search"
    params = {
        "q": f"{district}, {state}, {country}",
        "format": "json",
        "limit": 1
    }
    headers = {"User-Agent": "AgriVisionAI/1.0"}
    try:
        resp = requests.get(url, params=params, headers=headers, timeout=10)
        data = resp.json()
        if data:
            return float(data[0]['lat']), float(data[0]['lon'])
    except Exception as e:
        print(f"Error for {district}, {state}: {e}")
    return None, None

# test on first 5 before running all 311
for i in range(5):
    d, s = districts.loc[i, 'dist_name'], districts.loc[i, 'state_name']
    lat, lon = geocode_district(d, s)
    print(d, s, "->", lat, lon)
    time.sleep(1)  # Nominatim requires max 1 request/sec
```

```python
def extract_annual_values(param_dict, start_year=1981, end_year=2017):
    annual = {}
    for year in range(start_year, end_year + 1):
        key = f"{year}13"  # the 13th "month" = annual average
        if key in param_dict:
            annual[year] = param_dict[key]
    return annual

t2m_annual = extract_annual_values(params_data['T2M'])
print(list(t2m_annual.items())[:5])
print("Total years captured:", len(t2m_annual))
```

```python
def get_district_weather(lat, lon, start_year=1981, end_year=2017):
    url = "https://power.larc.nasa.gov/api/temporal/monthly/point"
    params = {
        "parameters": "T2M,RH2M,PRECTOTCORR,WS2M,ALLSKY_SFC_SW_DWN",
        "community": "AG",
        "longitude": lon,
        "latitude": lat,
        "start": start_year,
        "end": end_year,
        "format": "JSON"
    }
    try:
        resp = requests.get(url, params=params, timeout=30)
        data = resp.json()
        param_data = data['properties']['parameter']
        
        rows = []
        for year in range(start_year, end_year + 1):
            key = f"{year}13"
            row = {
                'year': year,
                'temperature_c': param_data['T2M'].get(key),
                'humidity_pct': param_data['RH2M'].get(key),
                'rainfall_mm': param_data['PRECTOTCORR'].get(key),
                'wind_speed_m_s': param_data['WS2M'].get(key),
                'solar_radiation_mj_m2_day': param_data['ALLSKY_SFC_SW_DWN'].get(key)
            }
            rows.append(row)
        return rows
    except Exception as e:
        print(f"Error: {e}")
        return None
```

```python
def get_district_weather(lat, lon, start_year=1984, end_year=2017):
    url = "https://power.larc.nasa.gov/api/temporal/monthly/point"
    params = {
        "parameters": "T2M,RH2M,PRECTOTCORR,WS2M,ALLSKY_SFC_SW_DWN",
        "community": "AG",
        "longitude": lon,
        "latitude": lat,
        "start": start_year,
        "end": end_year,
        "format": "JSON"
    }
    try:
        resp = requests.get(url, params=params, timeout=30)
        data = resp.json()
        param_data = data['properties']['parameter']
        
        rows = []
        for year in range(start_year, end_year + 1):
            key = f"{year}13"
            row = {
                'year': year,
                'temperature_c': param_data['T2M'].get(key),
                'humidity_pct': param_data['RH2M'].get(key),
                'rainfall_mm': param_data['PRECTOTCORR'].get(key, 0) * 365,  # convert daily rate to annual total
                'wind_speed_m_s': param_data['WS2M'].get(key),
                'solar_radiation_mj_m2_day': param_data['ALLSKY_SFC_SW_DWN'].get(key)
            }
            rows.append(row)
        return rows
    except Exception as e:
        print(f"Error: {e}")
        return None
```

```python
import pandas as pd
import numpy as np
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, f1_score, classification_report

df_a = pd.read_csv("../data/processed/dataset3_clean.csv")
print(df_a.shape)
df_a.head()
```

```python
X_a = df_a.drop(columns=['crop'])
y_a = df_a['crop']

le = LabelEncoder()
y_a_encoded = le.fit_transform(y_a)
print("Classes:", le.classes_)
print("Number of classes:", len(le.classes_))

X_train_a, X_test_a, y_train_a, y_test_a = train_test_split(
    X_a, y_a_encoded, test_size=0.2, random_state=42, stratify=y_a_encoded
)

scaler_a = StandardScaler()
X_train_a_scaled = scaler_a.fit_transform(X_train_a)
X_test_a_scaled = scaler_a.transform(X_test_a)

print(X_train_a_scaled.shape, X_test_a_scaled.shape)
```

```python
model_a = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(X_train_a_scaled.shape[1],)),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(22, activation='softmax')
])

model_a.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model_a.summary()
```

```python
history_a = model_a.fit(
    X_train_a_scaled, y_train_a,
    validation_split=0.15,
    epochs=100,
    batch_size=16,
    verbose=1,
    callbacks=[tf.keras.callbacks.EarlyStopping(patience=10, restore_best_weights=True)]
)
```

```python
y_pred_a_probs = model_a.predict(X_test_a_scaled)
y_pred_a = np.argmax(y_pred_a_probs, axis=1)

print("Test Accuracy:", accuracy_score(y_test_a, y_pred_a))
print("Test F1 (macro):", f1_score(y_test_a, y_pred_a, average='macro'))
print(classification_report(y_test_a, y_pred_a, target_names=le.classes_))
```

```python
import os
os.makedirs("../src/models/saved", exist_ok=True)

model_a.save("../src/models/saved/model_a_crop_recommendation.keras")
print("Model A saved")

# also save the scaler and label encoder — needed at inference time
import joblib
joblib.dump(scaler_a, "../src/models/saved/scaler_a.pkl")
joblib.dump(le, "../src/models/saved/label_encoder_a.pkl")
print("Scaler and label encoder saved")
```

```python
import pandas as pd
import numpy as np
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score

df_b = pd.read_csv("../data/processed/model_b_training_data.csv")
df_b_encoded = pd.get_dummies(df_b, columns=['state_name', 'crop', 'season'], drop_first=True)

X_b = df_b_encoded.drop(columns=['yield_kg_per_ha']).astype(float)
y_b = df_b_encoded['yield_kg_per_ha'].astype(float)

X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
    X_b, y_b, test_size=0.2, random_state=42
)

scaler_b = StandardScaler()
X_train_b_scaled = scaler_b.fit_transform(X_train_b)
X_test_b_scaled = scaler_b.transform(X_test_b)

print(X_train_b_scaled.shape, X_test_b_scaled.shape)
```

```python
model_b = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(X_train_b_scaled.shape[1],)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(1)
])

model_b.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

model_b.summary()
```

```python
history_b = model_b.fit(
    X_train_b_scaled, y_train_b,
    validation_split=0.15,
    epochs=100,
    batch_size=64,
    verbose=1,
    callbacks=[tf.keras.callbacks.EarlyStopping(patience=10, restore_best_weights=True)]
)
```

```python
y_pred_b = model_b.predict(X_test_b_scaled).flatten()
rmse_b = np.sqrt(mean_squared_error(y_test_b, y_pred_b))
r2_b = r2_score(y_test_b, y_pred_b)
print("Test RMSE:", rmse_b)
print("Test R²:", r2_b)
```

```python
y_train_b_log = np.log1p(y_train_b)
y_test_b_log = np.log1p(y_test_b)

model_b_log = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(X_train_b_scaled.shape[1],)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(1)
])

model_b_log.compile(optimizer='adam', loss='mse', metrics=['mae'])

history_b_log = model_b_log.fit(
    X_train_b_scaled, y_train_b_log,
    validation_split=0.15,
    epochs=100,
    batch_size=64,
    verbose=1,
    callbacks=[tf.keras.callbacks.EarlyStopping(patience=10, restore_best_weights=True)]
)
```

```python
y_pred_log = model_b_log.predict(X_test_b_scaled).flatten()
y_pred_b_final = np.expm1(y_pred_log)

rmse_log = np.sqrt(mean_squared_error(y_test_b, y_pred_b_final))
r2_log = r2_score(y_test_b, y_pred_b_final)
print("Test RMSE (log-transformed target):", rmse_log)
print("Test R² (log-transformed target):", r2_log)
```

```python
import pandas as pd
import numpy as np
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import mean_squared_error, r2_score

df_b = pd.read_csv("../data/processed/model_b_training_data.csv")

# Label encode categoricals (embeddings need integer indices, not one-hot)
le_state = LabelEncoder()
le_crop = LabelEncoder()
le_season = LabelEncoder()

df_b['state_idx'] = le_state.fit_transform(df_b['state_name'])
df_b['crop_idx'] = le_crop.fit_transform(df_b['crop'])
df_b['season_idx'] = le_season.fit_transform(df_b['season'])

print("States:", len(le_state.classes_))
print("Crops:", len(le_crop.classes_))
print("Seasons:", len(le_season.classes_))
```

```python
numeric_cols = ['area_ha', 'temperature_c', 'humidity_pct', 'rainfall_mm', 
                 'wind_speed_m_s', 'solar_radiation_mj_m2_day']

X_numeric = df_b[numeric_cols].values
X_state = df_b['state_idx'].values
X_crop = df_b['crop_idx'].values
X_season = df_b['season_idx'].values
y = df_b['yield_kg_per_ha'].values

# split all arrays together using the same indices
indices = np.arange(len(df_b))
train_idx, test_idx = train_test_split(indices, test_size=0.2, random_state=42)

X_num_train, X_num_test = X_numeric[train_idx], X_numeric[test_idx]
X_state_train, X_state_test = X_state[train_idx], X_state[test_idx]
X_crop_train, X_crop_test = X_crop[train_idx], X_crop[test_idx]
X_season_train, X_season_test = X_season[train_idx], X_season[test_idx]
y_train, y_test = y[train_idx], y[test_idx]

# scale only numeric features
scaler = StandardScaler()
X_num_train_scaled = scaler.fit_transform(X_num_train)
X_num_test_scaled = scaler.transform(X_num_test)

print(X_num_train_scaled.shape, X_state_train.shape, X_crop_train.shape, X_season_train.shape)
```

```python
# Embedding dimension: a common rule of thumb is min(50, (num_categories+1)//2)
state_emb_dim = min(10, (20+1)//2)  # 10
crop_emb_dim = min(4, (4+1)//2)      # 2
season_emb_dim = 2                    # only 2 categories, keep it small

# Inputs
numeric_input = tf.keras.layers.Input(shape=(6,), name='numeric_input')
state_input = tf.keras.layers.Input(shape=(1,), name='state_input')
crop_input = tf.keras.layers.Input(shape=(1,), name='crop_input')
season_input = tf.keras.layers.Input(shape=(1,), name='season_input')

# Embeddings
state_emb = tf.keras.layers.Embedding(input_dim=20, output_dim=state_emb_dim)(state_input)
state_emb = tf.keras.layers.Flatten()(state_emb)

crop_emb = tf.keras.layers.Embedding(input_dim=4, output_dim=crop_emb_dim)(crop_input)
crop_emb = tf.keras.layers.Flatten()(crop_emb)

season_emb = tf.keras.layers.Embedding(input_dim=2, output_dim=season_emb_dim)(season_input)
season_emb = tf.keras.layers.Flatten()(season_emb)

# Concatenate everything
concat = tf.keras.layers.Concatenate()([numeric_input, state_emb, crop_emb, season_emb])

# Dense layers
x = tf.keras.layers.Dense(128, activation='relu')(concat)
x = tf.keras.layers.Dropout(0.3)(x)
x = tf.keras.layers.Dense(64, activation='relu')(x)
x = tf.keras.layers.Dropout(0.2)(x)
x = tf.keras.layers.Dense(32, activation='relu')(x)
output = tf.keras.layers.Dense(1)(x)

model_b_emb = tf.keras.Model(
    inputs=[numeric_input, state_input, crop_input, season_input],
    outputs=output
)

model_b_emb.compile(optimizer='adam', loss='mse', metrics=['mae'])
model_b_emb.summary()
```

```python
history_b_emb = model_b_emb.fit(
    [X_num_train_scaled, X_state_train, X_crop_train, X_season_train],
    y_train,
    validation_split=0.15,
    epochs=100,
    batch_size=64,
    verbose=1,
    callbacks=[tf.keras.callbacks.EarlyStopping(patience=10, restore_best_weights=True)]
)
```

```python
y_pred_emb = model_b_emb.predict([X_num_test_scaled, X_state_test, X_crop_test, X_season_test]).flatten()
rmse_emb = np.sqrt(mean_squared_error(y_test, y_pred_emb))
r2_emb = r2_score(y_test, y_pred_emb)
print("Test RMSE (embeddings):", rmse_emb)
print("Test R² (embeddings):", r2_emb)
```

```python
y_pred_emb = model_b_emb.predict([X_num_test_scaled, X_state_test, X_crop_test, X_season_test]).flatten()
rmse_emb = np.sqrt(mean_squared_error(y_test, y_pred_emb))
r2_emb = r2_score(y_test, y_pred_emb)
print("Test RMSE (embeddings):", rmse_emb)
print("Test R² (embeddings):", r2_emb)
```
